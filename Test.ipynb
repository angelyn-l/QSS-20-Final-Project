```python
bb = df.groupby("music_id")["view_count"].mean()
```

```python
bb = bb.reset_index()
c = bb.sort_values("view_count", ascending=False)
d = c.view_count.value_counts().reset_index()
```

```python
plt.scatter(d.view_count, d["count"])
plt.yscale("log")
plt.xscale("log")

plt.xlabel("Views")
plt.ylabel("Number of Videos")
```

```python
plt.scatter(d["count"], d.view_count)

plt.ylabel("Views")
plt.xlabel("Number of Videos")
plt.xscale("log")
plt.yscale("log")
plt.show()
```
