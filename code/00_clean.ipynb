```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

## Functions:
# get_urls(music_id): Lists top 10 tracks by view count for each of the 100 soundtracks used by the most videos.
```

```python
# Load CSVs
df = pd.read_csv("master_tiktok_df.csv",
                dtype= {"music_id":str})

political_align = pd.read_csv("tiktok_poli_pref_results.csv")

top_100_tracks = pd.read_csv("data/top_100_tracks.csv",
                dtype= {"music_id":str})
```

```python
df = df.drop_duplicates("tiktokurl")

print("Size of dataset: " + str(df.shape))

top_100 = df.music_id.value_counts().iloc[0:100].reset_index(name = "counts")
```

```python
# Lists top 10 tracks by view count for each of the 100 soundtracks used by the most videos. 
# Each soundtrack was then hand-labeled for artist, title, and political leaning.

def get_urls(music_id):
    print(music_id)
    print(df[df.music_id == str(music_id)].sort_values(by = "view_count", ascending = False).tiktokurl.iloc[0:10])
    
pd.set_option('display.max_colwidth', None)
for x in top_100.music_id:
    get_urls(x)
```

```python
# Cleaning the dataset, standardizing column names, creating new columns for later analysis

top_100_tracks.columns = top_100_tracks.columns.str.lower()
top_100_tracks.columns = top_100_tracks.columns.str.replace(" ", "_")

top_100_tracks = top_100_tracks[~top_100_tracks.link.str.contains("Not available")]
top_100_tracks = top_100_tracks[top_100_tracks.is_music]

top_100_tracks["is_track_republican"] = top_100_tracks.track_category.str.contains("Republican").astype(int)
top_100_tracks["is_track_democratic"] = top_100_tracks.track_category.str.contains("Democratic").astype(int)


top_100_tracks["is_content_republican"] = top_100_tracks.content_category.str.contains("Republican").astype(int)
top_100_tracks["is_content_democratic"] = top_100_tracks.content_category.str.contains("Democratic").astype(int)


top_100_tracks = top_100_tracks[~top_100_tracks.content_category.str.contains("Neutral")]

top_100_df = df[df.music_id.isin(top_100_tracks.music_id)]
top_100_df = top_100_df[~top_100_df.video_description.str.contains("harrison", case = False, na = False) ]

print("Size of dataset: " + str(df.shape))
print("Size of dataset of TikToks using only top 100 tracks: " + str(top_100_df.shape))

# Prints number of rows before the merge
print("top_100_df rows (before merge):", len(top_100_df))
print("top_100_tracks rows:", len(top_100_tracks))

top_100_df = top_100_df.merge(top_100_tracks, 
                              left_on = "music_id", 
                              right_on = "music_id", 
                              how = "left")

# Prints number of rows after the merge
print("top_100_df rows (after merge):", len(top_100_df))
```

```python
top_with_politics = top_100_df.merge(political_align, 
                                     left_on = "id", 
                                     right_on = "id", 
                                     how = "inner")

# Prints number of rows before and after the merge
print("top_100_df rows (before merge):", len(top_100_df))
print("political_align rows (before merge):", len(political_align))
print("top_with_politics rows (after merge):", len(top_with_politics))

# Derive neutral-content political lean from caption/text flags
# Only applies to rows where the track itself isn't tagged R/D (is_track_* == 0)

top_with_politics["is_neutral_content_democratic"] = 0.0
top_with_politics["is_neutral_content_republican"] = 0.0

track_neutral_mask = (
    (top_with_politics["is_track_republican"] == 0) &
    (top_with_politics["is_track_democratic"] == 0)
)

dem_leaning_cols = ["pro_democrat", "against_republican", "pro_biden", "against_trump", "pro_kamala"]
rep_leaning_cols = ["against_democrat", "pro_republican", "against_biden", "pro_trump", "against_kamala"]

is_dem_leaning = top_with_politics[dem_leaning_cols].eq(1).any(axis=1)
is_rep_leaning = top_with_politics[rep_leaning_cols].eq(1).any(axis=1)

top_with_politics.loc[track_neutral_mask & is_dem_leaning, "is_neutral_content_democratic"] = 1.0
top_with_politics.loc[track_neutral_mask & is_rep_leaning, "is_neutral_content_republican"] = 1.0

# Filter out rows where both flags are 1 (ambiguous) or both are 0 (no signal)
both_flags = (
    top_with_politics["is_neutral_content_democratic"] +
    top_with_politics["is_neutral_content_republican"]
)

top_with_politics = top_with_politics[(both_flags == 1) | ~(top_with_politics.track_category.str.contains("Neutral"))].copy()

print("Number of TikToks in top_with_politics dataframe using neutral soundtracks: " + str(len(top_with_politics[(top_with_politics.is_track_democratic == 0.0) & (top_with_politics.is_track_republican == 0.0)])))
print("Number of TikToks in top_with_politics dataframe using pro-Democrat soundtracks: " + str(len(top_with_politics[(top_with_politics.is_track_democratic == 1.0) & (top_with_politics.is_track_republican == 0.0)])))
print("Number of TikToks in top_with_politics dataframe using pro-Republican soundtracks: " + str(len(top_with_politics[(top_with_politics.is_track_democratic == 0.0) & (top_with_politics.is_track_republican == 1.0)])))
print("Size of top_with_politics dataset: " + str(len(top_with_politics)))

# Save files

top_100_df.to_csv("data/top_100_df.csv", index=False)
top_100_tracks.to_csv("data/top_100_tracks.csv", index=False)
top_with_politics.to_csv("data/top_with_politics.csv", index=False)
```
